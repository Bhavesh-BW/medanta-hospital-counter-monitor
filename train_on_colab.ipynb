```python
# 3. Upload dataset.zip
# Click 'Choose Files' to upload dataset.zip from your computer
from google.colab import files
uploaded = files.upload()
```

```python
# 4. Unzip the dataset into /content/dataset
!unzip -q dataset.zip -d /content/

# Verify data.yaml
!cat /content/dataset/data.yaml
```

```python
# 5. Train YOLO11s on T4 GPU
from ultralytics import YOLO

# Load base model (yolo11s.pt will auto-download)
model = YOLO('yolo11s.pt')

# Start training with identical hyperparameters used by Medanta Counter Monitor
results = model.train(
    data='/content/dataset/data.yaml',
    epochs=50,
    imgsz=1280,
    batch=8,
    device=0,
    optimizer='AdamW',
    name='medanta_phone_detector',
    save=True,
    plots=True
)
```

```python
# 6. Download the resulting best.pt to your computer
from google.colab import files
files.download('runs/detect/medanta_phone_detector/weights/best.pt')
```
